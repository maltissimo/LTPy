```python
from pypylon import pylon

# 1. Connect to camera
camera = pylon.InstantCamera(pylon.TlFactory.GetInstance().CreateFirstDevice())
camera.Open()
nodemap = camera.GetNodeMap()

print("--- Initial State ---")
if hasattr(camera, "ExposureAuto"):
    print(f"ExposureAuto: {camera.ExposureAuto.Value}")
if hasattr(camera, "GainAuto"):
    print(f"GainAuto:     {camera.GainAuto.Value}")
if hasattr(camera, "GammaEnable"):
    print(f"GammaEnable:  {camera.GammaEnable.Value}")
```

```python
camera.Close()
```

```python

camera = pylon.InstantCamera(pylon.TlFactory.GetInstance().CreateFirstDevice())
camera.Open()
nodemap = camera.GetNodeMap()

print("--- Camera Model & Node Discovery ---")
print("Model:", camera.GetDeviceInfo().GetModelName())

# --- 1. Gamma Handling ---
# Case A: Camera uses GammaSelector (e.g., 'User', 'sRGB')
if hasattr(camera, "GammaSelector") and camera.GammaSelector.IsWritable():
    try:
        camera.GammaSelector.SetValue("User")
        print("GammaSelector set to: User")
    except Exception as e:
        print(f"GammaSelector error: {e}")

# Case B: Direct Gamma float node
if hasattr(camera, "Gamma") and camera.Gamma.IsWritable():
    camera.Gamma.SetValue(1.0)
    print(f"Gamma set directly to: {camera.Gamma.Value}")
else:
    # Check if node exists in nodemap but isn't a direct attribute
    gamma_node = nodemap.GetNode("Gamma")
    if gamma_node and pylon.IsWritable(gamma_node):
        gamma_node.SetValue(1.0)
        print(f"Gamma node set to: {gamma_node.GetValue()}")
    else:
        print("Gamma node: Not present or not writable (fixed hardware response).")
```
